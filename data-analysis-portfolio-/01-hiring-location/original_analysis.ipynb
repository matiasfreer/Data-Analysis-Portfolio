```python
df.loc[df['Job Title'].str.contains('data science', case=False, na=False), 'Job Title'] = 'Data Scientist'
df.loc[df['Job Title'].str.contains('data scientist', case=False, na=False), 'Job Title'] = 'Data Scientist'
df.loc[df['Job Title'].str.contains('data analyst', case=False, na=False), 'Job Title'] = 'Data Analyst'
df.loc[df['Job Title'].str.contains('data engineer', case=False, na=False), 'Job Title'] = 'Data Engineer'
df.loc[df['Job Title'].str.contains('machine learning engineer', case=False, na=False), 'Job Title'] = 'Machine Learning Engineer'
df.loc[df['Job Title'].str.contains('business intelligence analyst', case=False, na=False), 'Job Title'] = 'Business Intelligence Analyst'

# Standardize other 'Scientist' roles to 'Scientist', ensuring not to overwrite specific roles
mask_other_scientist = df['Job Title'].str.contains('scientist', case=False, na=False) & \
                       (df['Job Title'] != 'Data Scientist') & \
                       (df['Job Title'] != 'Data Analyst') & \
                       (df['Job Title'] != 'Data Engineer') & \
                       (df['Job Title'] != 'Machine Learning Engineer') & \
                       (df['Job Title'] != 'Business Intelligence Analyst')
df.loc[mask_other_scientist, 'Job Title'] = 'Scientist'

# Clean job titles by taking the part before '-', '/', or '('
def clean_job_title_separator(title):
    if pd.isna(title):
        return title
    # Find the index of the first '-', '/', or '('
    min_idx = len(title)
    if '-' in title:
        min_idx = min(min_idx, title.find('-'))
    if '/' in title:
        min_idx = min(min_idx, title.find('/'))
    if '(' in title:
        min_idx = min(min_idx, title.find('('))

    if min_idx < len(title):
        return title[:min_idx].strip()
    return title

df['Job Title'] = df['Job Title'].apply(clean_job_title_separator)

print("Unique Job Titles after standardization:")
display(df['Job Title'].unique())
```

```python
df['Salary Estimate'] = df['Salary Estimate'].apply(lambda x: x.split('(')[0] if pd.notna(x) else x)
display(df['Salary Estimate'].head())
```

Now, let's remove '$', 'K', and 'per hour' and convert the column to numeric values.

```python
df['Salary Estimate'] = df['Salary Estimate'].apply(lambda x: x.replace('$', '').replace('K', '').replace('per hour', '') if pd.notna(x) else x)
display(df['Salary Estimate'].head())
```

Next, we will split the 'Salary Estimate' into 'min_salary' and 'max_salary' columns and convert them to numeric type. We will handle hourly salaries by multiplying by a standard factor (e.g., 2080 hours per year) to convert them to annual salaries.

```python
df['min_salary'] = df['Salary Estimate'].apply(lambda x: int(x.split('-')[0]) * 1000 if pd.notna(x) and 'per hour' not in x and '-' in x else (int(x.split('per hour')[0]) * 2080 if pd.notna(x) and 'per hour' in x else (int(x) * 1000 if pd.notna(x) and '-' not in x and 'per hour' not in x else np.nan)))
df['max_salary'] = df['Salary Estimate'].apply(lambda x: int(x.split('-')[1]) * 1000 if pd.notna(x) and 'per hour' not in x and '-' in x else (int(x.split('per hour')[0]) * 2080 if pd.notna(x) and 'per hour' in x else (int(x) * 1000 if pd.notna(x) and '-' not in x and 'per hour' not in x else np.nan)))

df['min_salary'] = df.apply(lambda row: row['min_salary'] * 2 if 'per hour' in str(row['Salary Estimate']) else row['min_salary'], axis=1)
df['max_salary'] = df.apply(lambda row: row['max_salary'] * 2 if 'per hour' in str(row['Salary Estimate']) else row['max_salary'], axis=1)

# Adjust for values that were not ranges (single number, treated as both min and max originally)
df['max_salary'] = df.apply(lambda row: row['min_salary'] if pd.isna(row['max_salary']) and pd.notna(row['min_salary']) else row['max_salary'], axis=1)

df['max_salary'] = df['max_salary'].astype(float)
df['min_salary'] = df['min_salary'].astype(float)

display(df[['Salary Estimate', 'min_salary', 'max_salary']].head())
```

Now, let's calculate the average salary from 'min_salary' and 'max_salary' and create a new 'avg_salary' column.

```python
df['avg_salary'] = (df['min_salary'] + df['max_salary']) / 2
display(df[['min_salary', 'max_salary', 'avg_salary']].head())
```

```python
# 1. Check which states meet the Graph 1 screening criteria
# Criteria shown on the graph:
# 8–40 postings AND average compensation below $125,000

state_summary = (
    df_cleaned
    .dropna(subset=['state', 'avg_salary'])
    .groupby('state')
    .agg(
        job_postings=('state', 'size'),
        avg_compensation=('avg_salary', 'mean'),
        unique_employers=('Company Name', 'nunique')
    )
    .reset_index()
)

screened_states = state_summary[
    (state_summary['job_postings'] >= 8) &
    (state_summary['job_postings'] <= 40) &
    (state_summary['avg_compensation'] < 125000)
].sort_values('unique_employers', ascending=False)

print(screened_states)
```

### Evaluating States Based on Criteria

```python
display(screened_states)
```

Now, let's analyze these screened states. We want to identify the strongest recommendation for the AI gaming startup, considering:

*   **Average Compensation**: Lower is better (affordability).
*   **Job Postings**: Sufficient activity (market size).
*   **Unique Employers**: Fewer is better (less competition).

```python
# Sort states for better comparison based on the criteria
# Higher unique employers implies more competition, so we want to prioritize lower unique employers
# Lower avg_compensation is better
# Job postings between 8-40 are all considered 'sufficient' so we can look for states towards the middle of this range

# Let's create a composite score or rank them based on each criterion
# Assign ranks for each criterion (lower rank is better)
screened_states['comp_rank'] = screened_states['avg_compensation'].rank(ascending=True)
screened_states['employer_rank'] = screened_states['unique_employers'].rank(ascending=True)

# For job postings, being in the middle of the range (8-40) is ideal.
# Let's consider job postings closer to the middle of the range (e.g., around 20-25) as more balanced.
# For simplicity, we can initially rank them by job_postings and then consider the range.
# Or, given the range constraint is already applied, we might prioritize states with fewer postings if competition is a major concern,
# or more postings if market activity is a major concern.
# For now, let's just rank based on the other two and then discuss job postings.

# Let's consider a simple sum of ranks for now, giving equal weight. Adjust as needed.
screened_states['total_rank'] = screened_states['comp_rank'] + screened_states['employer_rank']

# Display states sorted by total rank
display(screened_states.sort_values('total_rank'))

print("\nDetailed comparison of screened states:")
for index, row in screened_states.sort_values('total_rank').iterrows():
    state = row['state']
    postings = row['job_postings']
    avg_comp = row['avg_compensation']
    employers = row['unique_employers']
    print(f"State: {state} | Postings: {postings} | Avg Comp: ${avg_comp:,.2f} | Unique Employers: {employers}")
```

### Analysis and Recommendation

Based on the data and the established criteria:

*   **Affordability (Average Compensation):** Lower average compensation is preferred. States like TN, CO, GA, MO have the lowest average salaries within the screened group.
*   **Job Posting Activity (8-40 postings):** All screened states meet this criterion, indicating a sufficient level of market activity. States with fewer postings (e.g., FL, TN, GA) might have less competition from other companies in terms of sheer volume, but also less overall activity.
*   **Employer Competition (Unique Employers):** Fewer unique employers are preferred. States like FL, TN, CO, GA have the fewest unique employers, suggesting less competition for talent.

Let's consider **Colorado (CO)**:
*   **Average Compensation:** $103,650.00 (among the lowest).
*   **Job Postings:** 10 (meets the criteria, lower end of the range, suggesting manageable activity).
*   **Unique Employers:** 6 (among the lowest, indicating less competition).

Compared to other states, Colorado stands out for its very competitive average compensation and low number of unique employers, while still maintaining a sufficient number of job postings. Other states like Tennessee (TN) also offer low average compensation and unique employers but have similar job posting counts.

**Limitations and Trade-offs:**

1.  **Limited Data Range:** The dataset covers a specific period and might not reflect long-term trends or seasonal variations in job postings and salaries.
2.  **Proxy for Competition:** The number of unique employers is a proxy. A high number could mean more options for candidates, but also more competition for the startup. A low number *could* mean less competition, but also a smaller talent pool or less dynamic market. We are treating it as less competition for the startup.
3.  **No Talent Supply Data:** The analysis explicitly avoids interpreting job postings as direct evidence of available talent supply. A state might have few postings but a large pool of available data scientists.
4.  **Job Title Standardization:** While job titles were standardized, some nuances might be lost, potentially affecting the accuracy of job posting counts.
5.  **Small Sample Size:** For some states, the number of job postings is at the lower end of the acceptable range (e.g., 8-10 postings), which might make the average compensation and employer competition less robustly representative.

### Conclusion:

Based strictly on this dataset and these criteria, **Colorado (CO)** provides the strongest overall balance for the startup. It offers one of the lowest average compensations, a manageable number of job postings indicating an active but not overly saturated market, and a low number of unique employers, suggesting reduced recruiting competition. While other states like Tennessee (TN) and Georgia (GA) also perform well in terms of affordability and employer competition, Colorado's specific balance appears slightly more favorable within the given constraints.
